```python
from google.colab import drive
drive.mount('/content/drive')
import os

songs_dir = "/content/drive/MyDrive/Dataset/metal"

import re

# List MP3 files
files = [f for f in os.listdir(songs_dir) if f.lower().endswith(".mp3")]
files.sort()

# Sequential numbering
for idx, old_name in enumerate(files, start=1):
    new_name = f"Metal_{idx:02d}.mp3"

    old_path = os.path.join(songs_dir, old_name)
    new_path = os.path.join(songs_dir, new_name)

    os.rename(old_path, new_path)
    print(f"Renamed: {old_name} → {new_name}")
```
